# ⏰ Guia Completo do Tipo `datetime` em Python

## 📖 O que é `datetime`?

A classe `datetime` é um tipo de dado que representa **uma data e hora específicas** em Python.
Ela permite manipular:

* Ano, mês, dia
* Hora, minuto, segundo, microssegundo
* Dias da semana (segunda, terça, etc.)

A classe `datetime` faz parte do módulo **`datetime`**, que é padrão do Python e essencial para manipulação de datas e horários, cálculos de diferenças entre datas e formatação em diferentes padrões.

Em **análises de dados**, ela é muito usada para trabalhar com **séries temporais**.

---

## 🐼 `datetime` com Pandas

Com **Pandas**, podemos:

* Converter strings em objetos `datetime`
* Filtrar dados por intervalos de tempo
* Agregar dados por hora, dia, mês ou ano
* Realizar operações temporais de forma eficiente dentro de DataFrames

Exemplo prático:

```python
import pandas as pd

# criando um DataFrame com datas em string
df = pd.DataFrame({
    "data": ["2022-01-01", "2022-01-02", "2022-01-03"],
    "vendas": [100, 150, 200]
})

# convertendo para datetime
df['data'] = pd.to_datetime(df['data'])

# filtrando vendas após 2022-01-01
df_filtrado = df[df['data'] > '2022-01-01']
print(df_filtrado)
```

## 📚 Biblioteca `datetime` do Python

A biblioteca **`datetime`** fornece classes para trabalhar com datas e horas.

### 1️⃣ Criando um objeto `datetime` com a data e hora atual

```python
import datetime

# criando um objeto datetime com a data e hora atual
agora = datetime.datetime.now()

print("Data e hora atual:", agora)
```

* **`datetime.datetime.now()`** retorna um objeto `datetime` com **data e hora atual**.

---

### 2️⃣ Criando um objeto `date` com apenas a data

```python
import datetime

# criando um objeto date com a data de hoje
hoje = datetime.date.today()

print("Data de hoje:", hoje)
```

* **`datetime.date.today()`** retorna um objeto com **apenas a data atual**.

---

### 3️⃣ Operações matemáticas com datas

Você pode **calcular diferenças entre datas**:

```python
import datetime

# criando dois objetos date
data_1 = datetime.date(2022, 1, 1)
data_2 = datetime.date(2023, 1, 1)

# diferença entre datas
diferenca = data_2 - data_1

print("Diferença entre as datas:", diferenca)
```

* Resultado: `365 days` (diferença entre `data_2` e `data_1`)

Você também pode somar ou subtrair **`timedelta`** para manipular datas.

### Agrupando vagas por mês com `groupby`

Nosso objetivo é **agrupar os dados por mês** e calcular **quantas vagas estavam disponíveis em cada mês**. Para isso, usamos o método `groupby` do Pandas.

#### Passo a passo:

1. **Criar o agrupamento por mês**
   Usamos a coluna `data` para extrair o ano e o mês de cada registro. Com isso, todas as linhas do mesmo mês serão agrupadas:

```python
dt_data['data'].dt.strftime('%Y-%m')
```

2. **Aplicar o `groupby`**
   Passamos a seleção de "ano–mês" para dentro do `groupby()` e selecionamos a coluna `vaga_disponivel`:

```python
dt_data.groupby(dt_data['data'].dt.strftime('%Y-%m'))['vaga_disponivel']
```

3. **Somar os valores booleanos**
   Como a coluna `vaga_disponivel` contém valores booleanos (`True` ou `False`), o Python trata:

* `True` → 1
* `False` → 0

Assim, a função `sum()` contará **quantos dias houve vagas disponíveis** em cada mês:

```python
dt_data.groupby(dt_data['data'].dt.strftime('%Y-%m'))['vaga_disponivel'].sum()
```

4. **Resultado**
   O resultado será uma tabela com cada mês e a soma de vagas disponíveis naquele período. Por exemplo:

| data    | vaga\_disponivel |
| ------- | ---------------- |
| 2016-01 | 16543            |
| 2016-02 | 20128            |
| 2016-03 | 23357            |
| ...     | ...              |

🔹 **Interpretação:** Cada número indica **quantos dias naquele mês tinham vagas disponíveis**.

Nas colunas `datas_de_pagamento` e `datas_combinadas_pagamento`, temos datas em formato 'dia/mês/ano' (`dd/mm/AAAA`). Transforme esses dados para o tipo datetime e busque uma forma de visualização de subconjunto que possa contribuir no objetivo do contexto que os dados estão inseridos.

Se você não lembra o problema do projeto desafio 2, vou deixar abaixo o texto da situação para facilitar o encontro da informação:

>A administração de condomínios é uma tarefa que requer muita atenção e organização. Entre as diversas responsabilidades da gestão está o recebimento dos aluguéis dos locatários. Para garantir uma boa saúde financeira do empreendimento, é fundamental que esses pagamentos sejam feitos de forma regular e pontual. No entanto, sabemos que nem sempre isso acontece. Pensando nisso, proponho um desafio de tratamento de dados com o objetivo de analisar o atraso no pagamento de aluguéis no condomínio fictício de alguns moradores.

````python
import pandas as pd
import numpy as np
import io

def carregar_dados(url: str, chave: str, coluna_valor: str, prefixo_remover: list[str] = None) -> pd.DataFrame:
    """
    Lê um JSON de uma URL, normaliza os dados, explode listas e limpa a coluna de valores monetários.

    Args:
        url (str): Link do arquivo JSON.
        chave (str): Nome da chave principal do JSON (ex.: 'dados_vendas').
        coluna_valor (str): Nome da coluna que contém os valores monetários.
        prefixo_remover (list[str], opcional): Strings a serem removidas antes de converter em float.

    Returns:
        pd.DataFrame: DataFrame normalizado e com a coluna de valores em float.
    """
    # Carrega e normaliza
    dados = pd.read_json(io.StringIO(pd.read_json(url).to_json()))
    dados = pd.json_normalize(dados[chave])

    # Explode as colunas de listas
    colunas = list(dados.columns)
    dados = dados.explode(colunas[1:]).reset_index(drop=True)

    # Limpeza da coluna de valores
    if prefixo_remover:
        for item in prefixo_remover:
            dados[coluna_valor] = dados[coluna_valor].str.replace(item, '', regex=False)

    dados[coluna_valor] = dados[coluna_valor].str.replace(',', '.', regex=False).str.strip()
    dados[coluna_valor] = dados[coluna_valor].astype(np.float64)

    return dados


url_vendas = "https://raw.githubusercontent.com/YuriArduino/Estudos_Pandas/refs/heads/data-tests/dados_vendas_clientes.json"
url_locacao = "https://raw.githubusercontent.com/YuriArduino/Estudos_Pandas/refs/heads/data-tests/dados_locacao_imoveis.json"

dados_vendas = carregar_dados(url_vendas, chave="dados_vendas", coluna_valor="Valor da compra", prefixo_remover=["R$ "])
dados_locacao = carregar_dados(url_locacao, chave="dados_locacao", coluna_valor="valor_aluguel", prefixo_remover=["$", " reais"])

print(dados_vendas.head())
print(dados_locacao.head())
````

````python
dados_vendas['Cliente'] = dados_vendas['Cliente'].str.lower()
dados_vendas['Cliente'] = dados_vendas['Cliente'].str.replace(r'[^a-z]', ' ', regex=True)
dados_locacao['apartamento'] = dados_locacao['apartamento'].str.replace('(blocoAP)', '', regex=False)
````

````python
dados_vendas['Data de venda'] = pd.to_datetime(dados_vendas['Data de venda'], format='%d/%m/%Y')

total_compras = dados_vendas.groupby(['Cliente'])['Valor da compra'].sum()

total_compras
````

````python
dados_locacao['datas_combinadas_pagamento'] = pd.to_datetime(dados_locacao['datas_combinadas_pagamento'], format='%d/%m/%Y')
dados_locacao['datas_de_pagamento'] = pd.to_datetime(dados_locacao['datas_de_pagamento'], format='%d/%m/%Y')
dados_locacao['atraso'] = (dados_locacao['datas_de_pagamento'] - dados_locacao['datas_combinadas_pagamento']).dt.days

media_atraso = dados_locacao.groupby(['apartamento'])['atraso'].mean()

media_atraso
````
